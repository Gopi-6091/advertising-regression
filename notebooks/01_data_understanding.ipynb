```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="TV",
    bins=30,
    kde=True
)

plt.title("Distribution of TV Advertising Expenditure")
plt.xlabel("TV Advertising Expenditure")
plt.ylabel("Number of Markets")

plt.show()

```

```python
Q1 = df["TV"].quantile(0.25)
Q3 = df["TV"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(x=df["TV"])

plt.xticks(range(0, 301, 20))

plt.title("Boxplot of TV Advertising Expenditure")
plt.xlabel("TV Advertising Expenditure")

plt.show()
```

```python
Q1 = df["TV"].quantile(0.25)
Q3 = df["TV"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

plt.figure(figsize=(8, 4))

sns.boxplot(x=df["TV"])

plt.axvline(
    lower_bound,
    color="red",
    linestyle="--",
    label=f"Lower IQR: {lower_bound:.1f}"
)

plt.axvline(
    upper_bound,
    color="red",
    linestyle="--",
    label=f"Upper IQR: {upper_bound:.1f}"
)

plt.xticks(range(0, 301, 40))

plt.title("Boxplot of TV Advertising Expenditure")
plt.xlabel("TV Advertising Expenditure")
plt.legend()

plt.show()
```

```python
## Data Cleaning & Univariate Analysis — Summary

- No missing values, duplicate rows, or negative values were found.
- `Unnamed: 0` was identified as an observation identifier and will be excluded from modeling.
- Histograms + KDE were used to understand the distribution of each variable.
- TV appeared multimodal, radio relatively flat, newspaper right-skewed, and sales approximately bell-shaped with slight right skew.
- Potential outliers were investigated using the IQR method: `IQR = Q3 - Q1`.
- IQR boundaries were calculated as `Q1 - 1.5 × IQR` and `Q3 + 1.5 × IQR`.
- TV, radio, and sales had no IQR-based potential outliers.
- Newspaper had two potential outliers, which were investigated and found to be plausible observations.
- No observations were removed or modified; the raw dataset remains unchanged.
- Key principle: **Observe → Investigate → Reason → Decide → Document.**
```
