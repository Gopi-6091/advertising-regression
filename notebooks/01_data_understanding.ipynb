### Correlation Analysis

Correlation measures the **direction and strength of a linear relationship** between two numerical variables.

Pearson correlation (`r`) ranges from **-1 to +1**:

- `+1` → perfect positive linear relationship
- `0` → no linear relationship
- `-1` → perfect negative linear relationship

Conceptually, Pearson correlation measures whether two variables tend to move together relative to their respective means.

The formula is:

r = Σ[(xᵢ − x̄)(yᵢ − ȳ)] / √[Σ(xᵢ − x̄)² × Σ(yᵢ − ȳ)²]

For this project, the correlations with sales are:

- TV → Sales: **0.782**
- Radio → Sales: **0.576**
- Newspaper → Sales: **0.228**

This suggests that TV has the strongest positive linear association with sales, followed by radio, while newspaper has a much weaker positive linear association.

### Important: Correlation ≠ Causation

A high correlation does **not** prove that one variable causes another to change.

For example, the positive correlation between TV advertising and sales does not prove that increasing TV advertising causes sales to increase.

This dataset is observational, so correlation should be interpreted as **association**, not causation.

Correlation also measures **linear** association. A correlation close to zero does not necessarily mean that no relationship exists; a non-linear relationship may still be present.

### EDA Principle

We use correlation to **quantify patterns observed in the scatter plots**. It supports our exploratory analysis but does not by itself establish causality or determine the optimal advertising strategy.

### Simple Linear Regression — Quick Notes

Model:
Sales = β₀ + β₁(TV) + ε

- β₀ = intercept → predicted Sales when TV = 0
- β₁ = slope/coefficient → expected change in Sales for 1-unit increase in TV
- ε = error → variation in Sales not explained by the model
- β̂₁ = estimated coefficient from our sample
- β₁ = true population coefficient

Example:
Sales = 7 + 0.05(TV)

→ TV increases by 1 unit → predicted Sales increases by 0.05 units
→ TV = 10 → predicted Sales = 7.5

### Hypothesis Testing

Question:
Is the true population slope different from 0?

H₀: β₁ = 0
→ No linear association between TV and Sales

H₁: β₁ ≠ 0
→ Evidence of a non-zero linear association

Important:
- Correlation ≠ causation
- Statistical significance ≠ good prediction
- Hypothesis testing ≠ testing whether the relationship is linear
- Reject H₀ → evidence β₁ differs from 0
- Fail to reject H₀ → insufficient evidence that β₁ differs from 0

R²	Proportion of target variation explained by model	Unitless	Overall model fit

```python
from sklearn.model_selection import train_test_split

X = df[["TV", "radio", "newspaper"]]
y = df["sales"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

# Train the model
model.fit(X_train, y_train)

# Predictions
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)

for feature, coefficient in zip(X.columns, model.coef_):
    print(f"{feature}: {coefficient:.4f}")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Training metrics
train_mae = mean_absolute_error(y_train, y_train_pred)
train_mse = mean_squared_error(y_train, y_train_pred)
train_rmse = np.sqrt(train_mse)
train_r2 = r2_score(y_train, y_train_pred)

# Test metrics
test_mae = mean_absolute_error(y_test, y_test_pred)
test_mse = mean_squared_error(y_test, y_test_pred)
test_rmse = np.sqrt(test_mse)
test_r2 = r2_score(y_test, y_test_pred)

print("Training Performance")
print(f"MAE:  {train_mae:.3f}")
print(f"MSE:  {train_mse:.3f}")
print(f"RMSE: {train_rmse:.3f}")
print(f"R²:   {train_r2:.3f}")

print("\nTest Performance")
print(f"MAE:  {test_mae:.3f}")
print(f"MSE:  {test_mse:.3f}")
print(f"RMSE: {test_rmse:.3f}")
print(f"R²:   {test_r2:.3f}")
```
